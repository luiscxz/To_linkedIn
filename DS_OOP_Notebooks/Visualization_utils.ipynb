# Plot time series

Para crear una clase, es importante tener en cuenta lo siguiente:

- `Atributos` → Son las propiedades o características que describen a los objetos creados con la clase.

    - Los datos (por ejemplo, un DataFrame de pandas o una lista de valores de la serie de tiempo).

    - El título del gráfico.

    - El estilo (colores, grosor de línea, tipo de gráfico).

    - El tamaño de la figura.

- `Métodos` → Son las acciones o comportamientos que los objetos pueden realizar:

    - linea() → Para graficar la serie como línea.

    - area() → Para graficar como área.

    - show() → Para mostrar el gráfico en pantalla.


La clase está compuesta por lo siguiente:

1. Se define la clase y el nombre.
2. Se crea el constructor de la clase. En este caso el objeto es el graficador
3. Es el título que tendrá el gráfico.
4. Es el nombre que se le pondrá al eje x
5. Es el nombre que se le pondrá al eje y
6. Es el nombre de la variable que se va a graficar
7. Es nombre del color que usará en la linea que representa la serie temporal
8. Es el grosor de la linea
9. Es el tipo de linea: 
    - solid: Linea continua
    - dash: Linea discontinua - - -
    - dot: Linea punteada ...
    - dashdot: Linea -.-.
    - longdash: Linea discontinua Larga
10. Es la variable tiempo.
11. Es la variable que se quiere ver en el tiempo
12. Es la altura del gráfico
13. Es el largo del gráfico
14. Es la forma en como se mostraran las fechas en el eje X. 
15. Establece si se coloca area bajo la curva
    - none: Sin relleno
    - tozeroy: Rellena el área desde la línea hacia el eje Y=0.
    - tozerox: Rellena el área desde la línea hacia el eje X=0.
    - tonexty: Rellena el área entre esta línea y la anterior en el eje Y.
    - tonextx: Rellena el área entre esta línea y la anterior en el eje X.
    - toself: rellena el polígono formado por los puntos de la serie (cuando se cierra sobre sí mismo).
16. Establece el color de relleno y la opacidad.



```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

class VisualizarSerieTemporal: #1
    def __init__(self,x,y,titulo,xlabel,ylabel,nombreVariable,
                 LineColor,LineWidth,dash,altura,largo,x_tickformat, fill,fillcolor): #2
        # Inicializar el objeto graficador
        self.titulo = titulo  #3
        self.xlabel = xlabel  #4
        self.ylabel = ylabel  #5
        self.nombreVariable = nombreVariable #6
        self.LineColor = LineColor  #7
        self.LineWidth = LineWidth  #8
        self.dash = dash    #9
        self.x = x  #10
        self.y = y  #11
        self.altura = altura #12
        self.largo = largo #13
        self.x_tickformat = x_tickformat #14
        self.fill = fill    #15
        self.fillcolor = fillcolor      #16

    # Función que contiene el diseño base de la gráfica
    def diseno_base(self, fig):
        fig.update_layout(
        title=f'<b>{self.titulo}</b>',
        title_font=dict(size=20, family='Arial', color='black'),
        xaxis_title=f'<b>{self.xlabel}</b>',
        yaxis_title=f'<b>{self.ylabel}</b>',
        font=dict(family='Arial', size=12, color='black'),
        plot_bgcolor='white',
        paper_bgcolor='white',
        hovermode='x unified',
        height=self.altura,
        margin=dict(l=50, r=50, b=50, t=80)
        )
        # Formato año-mes en eje x
        fig.update_xaxes(showgrid=True, gridcolor='lightgray', tickformat=self.x_tickformat)
        fig.update_yaxes(showgrid=True, gridcolor='lightgray')
        return fig

    # Función para graficar una serie temporal con una linea.
    def GraficarUnaVariableTemporal(self):
        if not self.x.dtype.name.startswith('datetime'):
            self.x = pd.to_datetime(self.x)
        
        # Crear figura
        fig = go.Figure()
        fig.add_trace(go.Scatter(
            x=self.x,
            y=self.y,
            mode=self.dash,
            name=self.nombreVariable.capitalize(),
            line=dict(color=self.LineColor, width=self.LineWidth, dash=self.dash),
            fill=self.fill,
            fillcolor=self.fillcolor  
        ))

        # Aplicar diseño común
        fig = self.diseno_base(fig)
        return fig
    
    # Funcion para graficar la serie de tiempo, el incremento porcentual y el boxplot del incremento porcentual
    def GraficarVariableTemporalMasIncrementoPorcentual(self):
        # Crear la figura con subgráficos
        fig = make_subplots(
            rows=3, cols=1, shared_xaxes=True, vertical_spacing=0.1,
            row_heights=[0.5, 0.35, 0.15],
            specs=[[{"type": "scatter"}], [{"type": "scatter"}], [{"type": "box"}]]
        )

        y_mediamovil = self.y.rolling(window=3).mean()
        IncrementoPorcentual = 100 *abs((self.y - y_mediamovil)/y_mediamovil).round(2)
 
        color_cambio = '#D62728'  
        color_boxplot = '#2CA02C' 
        # Gráfico 1: Serie temporal 
        fig.add_trace(
            go.Scatter(
            x=self.x,
            y=self.y,
            mode='lines',
            name=self.nombreVariable.capitalize(),
            line=dict(color=self.LineColor, width=self.LineWidth, dash=self.dash),
            fill=self.fill,
            fillcolor=self.fillcolor  
        ),row=1, col=1)
        
        # Gráfico 2: Incremento Porcentual
        fig.add_trace(go.Scatter(
            x=self.x, 
            y=IncrementoPorcentual,
            mode='lines', 
            name='Incremento %', 
            line=dict(color=color_cambio, width=1.5),
            fill=self.fill,
            fillcolor='rgba(214, 39, 40, 0.2)'
        ), row=2, col=1)

        # Gráfico 3: Boxplot del Incremento Porcentual (horizontal)
        fig.add_trace(go.Box(
            x=IncrementoPorcentual, 
            name='',
            boxpoints='outliers',
            marker=dict(
                color=color_boxplot, 
                outliercolor='#7F7F7F',  # Gris para outliers
                line=dict(width=1.2)
            ),
            line=dict(color=color_boxplot)
        ), row=3, col=1)

        # Diseño 
        fig.update_layout(
        height=self.altura, 
        width=self.largo,
        title_text=self.titulo,
        title_font=dict(size=20, family="Arial Bold", color='#333333'),
        showlegend=True,
        legend=dict(
            orientation='h',
            yanchor='top',
            y=-0.15,
            xanchor='center',
            x=0.5,
            font=dict(color='#333333')
        ),
        font=dict(size=14, color='#333333'),  # Texto en gris oscuro
        margin=dict(l=70, r=20, t=50, b=70),
        plot_bgcolor='white',
        paper_bgcolor='white',
        )

        # Configuración de ejes
        for row in [1, 2, 3]:
            fig.update_yaxes(
                row=row, col=1,
                title_standoff=10,
                showgrid=True,
                gridcolor='rgba(200, 200, 200, 0.5)',  # Gris claro para la grilla
                title_font=dict(color='#333333'),
                tickfont=dict(color='#333333')
            )
            
        fig.update_xaxes(
            showgrid=True,
            gridcolor='rgba(200, 200, 200, 0.5)',
            tickformat=self.x_tickformat,
            title_font=dict(color='#333333'),
            tickfont=dict(color='#333333')
        )
            # Etiquetas de los ejes X
        fig.update_xaxes(title_text="", row=1, col=1, title_standoff=10, 
                    showgrid=True, gridcolor='lightgray', showticklabels=True)
        fig.update_xaxes(title_text="", row=2, col=1, title_standoff=10, 
                        showgrid=True, gridcolor='lightgray', showticklabels=True)

        # Formatear las fechas en los ejes X
        fig.update_xaxes(tickformat=self.x_tickformat, row=1, col=1)  # Formato de fecha para el gráfico 1
        fig.update_xaxes(tickformat=self.x_tickformat, row=2, col=1)  # Formato de fecha para el gráfico 2
        
        # Etiquetas específicas
        fig.update_yaxes(title_text=self.ylabel, row=1, col=1)
        fig.update_yaxes(title_text="Incre %", row=2, col=1)
        fig.update_yaxes(title_text="Boxplot", row=3, col=1)
        return fig

    
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

class VisualizarSerieTemporal:
    def __init__(self,x,y,titulo,xlabel,ylabel,nombreVariable,
                 LineColor="blue",LineWidth=2,dash="solid",mode="lines",
                 altura=600,largo=900,x_tickformat="%Y-%m-%d",
                 fill=None,fillcolor=None):

        # Inicializar atributos
        self.titulo = titulo    #3
        self.xlabel = xlabel    #4
        self.ylabel = ylabel    #5
        self.nombreVariable = nombreVariable    #6
        self.LineColor = LineColor  #7
        self.LineWidth = LineWidth  #8
        self.dash = dash            #9 tipo de línea: solid, dash, dot...
        self.mode = mode            # modo de gráfico: lines, markers, text...
        self.x = x  #10
        self.y = y  #11
        self.altura = altura    #12
        self.largo = largo  #13
        self.x_tickformat = x_tickformat    #14
        self.fill = fill    #15
        self.fillcolor = fillcolor  #15

    # Diseño base
    def diseno_base(self, fig):
        fig.update_layout(
            title=f'<b>{self.titulo}</b>',
            title_font=dict(size=20, family='Arial', color='black'),
            xaxis_title=f'<b>{self.xlabel}</b>',
            yaxis_title=f'<b>{self.ylabel}</b>',
            font=dict(family='Arial', size=12, color='black'),
            plot_bgcolor='white',
            paper_bgcolor='white',
            hovermode='x unified',
            height=self.altura,
            width=self.largo,
            margin=dict(l=50, r=50, b=50, t=80)
        )
        fig.update_xaxes(showgrid=True, gridcolor='lightgray', tickformat=self.x_tickformat)
        fig.update_yaxes(showgrid=True, gridcolor='lightgray')
        return fig

    # Serie temporal simple
    def GraficarUnaVariableTemporal(self):
        if not self.x.dtype.name.startswith('datetime'):
            self.x = pd.to_datetime(self.x)

        fig = go.Figure()
        fig.add_trace(go.Scatter(
            x=self.x,
            y=self.y,
            mode=self.mode,   # ahora controlado por separado
            name=self.nombreVariable.capitalize(),
            line=dict(color=self.LineColor, width=self.LineWidth, dash=self.dash),
            fill=self.fill,
            fillcolor=self.fillcolor  
        ))

        return self.diseno_base(fig)

    # Serie temporal + incremento % + boxplot
    def GraficarVariableTemporalMasIncrementoPorcentual(self):
        fig = make_subplots(
            rows=3, cols=1, shared_xaxes=True, vertical_spacing=0.1,
            row_heights=[0.5, 0.35, 0.15],
            specs=[[{"type": "scatter"}], [{"type": "scatter"}], [{"type": "box"}]]
        )

        y_mediamovil = self.y.rolling(window=3).mean()
        IncrementoPorcentual = 100 * abs((self.y - y_mediamovil)/y_mediamovil).round(2)

        color_cambio = '#D62728'  
        color_boxplot = '#2CA02C' 

        # Serie temporal
        fig.add_trace(
            go.Scatter(
                x=self.x,
                y=self.y,
                mode=self.mode,
                name=self.nombreVariable.capitalize(),
                line=dict(color=self.LineColor, width=self.LineWidth, dash=self.dash),
                fill=self.fill,
                fillcolor=self.fillcolor  
            ), row=1, col=1)

        # Incremento porcentual
        fig.add_trace(go.Scatter(
            x=self.x, 
            y=IncrementoPorcentual,
            mode="lines", 
            name="Incremento %",
            line=dict(color=color_cambio, width=1.5),
            fill=self.fill,
            fillcolor='rgba(214, 39, 40, 0.2)'
        ), row=2, col=1)

        # Boxplot incremento %
        fig.add_trace(go.Box(
            x=IncrementoPorcentual, 
            name='',
            boxpoints='outliers',
            marker=dict(
                color=color_boxplot, 
                outliercolor='#7F7F7F',
                line=dict(width=1.2)
            ),
            line=dict(color=color_boxplot)
        ), row=3, col=1)

        # Layout
        fig.update_layout(
            height=self.altura, 
            width=self.largo,
            title_text=self.titulo,
            title_font=dict(size=20, family="Arial Bold", color='#333333'),
            showlegend=True,
            legend=dict(
                orientation='h',
                yanchor='top',
                y=-0.15,
                xanchor='center',
                x=0.5,
                font=dict(color='#333333')
            ),
            font=dict(size=14, color='#333333'),
            margin=dict(l=70, r=20, t=50, b=70),
            plot_bgcolor='white',
            paper_bgcolor='white',
        )

        # Ejes
        for row in [1, 2, 3]:
            fig.update_yaxes(
                row=row, col=1,
                title_standoff=10,
                showgrid=True,
                gridcolor='rgba(200, 200, 200, 0.5)',
                title_font=dict(color='#333333'),
                tickfont=dict(color='#333333')
            )
            
        fig.update_xaxes(
            showgrid=True,
            gridcolor='rgba(200, 200, 200, 0.5)',
            tickformat=self.x_tickformat,
            title_font=dict(color='#333333'),
            tickfont=dict(color='#333333')
        )
        fig.update_xaxes(title_text="", row=1, col=1, showticklabels=True)
        fig.update_xaxes(title_text="", row=2, col=1, showticklabels=True)

        # Etiquetas específicas
        fig.update_yaxes(title_text=self.ylabel, row=1, col=1)
        fig.update_yaxes(title_text="Incre %", row=2, col=1)
        fig.update_yaxes(title_text="Boxplot", row=3, col=1)

        return fig
```

```python
grafico = VisualizarSerieTemporal(
    x=dataStore['Date'],
    y=dataStore[''],
    titulo="Serie Temporal de Temperatura",
    xlabel="Fecha",
    ylabel="Valor",
    nombreVariable="Temperatura",
    LineColor="blue",
    LineWidth=2,
    dash="solid",
    altura=600,
    largo=1000,
    x_tickformat="%Y-%m-%d",
    fill="tozeroy",   # activa relleno desde la línea hasta el eje x
    fillcolor="rgba(31, 119, 180, 0.2)" # opacidad del área
)
fig1 = grafico.GraficarVariableTemporalMasIncrementoPorcentual()
fig1.show()
```
